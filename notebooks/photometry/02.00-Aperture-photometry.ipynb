# Aperture photometry

Photometry techniques largely fall into one of two categories: aperture photometry and point spread function (PSF) photometry.

In aperture photometry, software is used to draw a shape, typically a circle or an ellipse, around the object of interest and pixel values within the aperture are summed to determine the "amount" of light within the aperture. Wheether that "amount" is a flux or counts or something else depends on what units the image is in. This short summary brushes over several complications, among them: how are pixels partially in the aperture handled? How is sky background removed? 

In PSF photometry, the assumption is that one has a good understanding of what analytic function fits the shape of a point source in the image, for example a two dimensional Gaussian. That function is fit to each source in the image and the fit parameters are used to determine the "amount" of light in the source. As with aperture photometry there are complications among them: How do you determine the appropriate PSF? What if the PSF varies across the detector, or varies from night to night (for a ground based telescope)?

In either case, the raw measurements made of the amount of light for each source needs additional calibration before it is useful for science.
