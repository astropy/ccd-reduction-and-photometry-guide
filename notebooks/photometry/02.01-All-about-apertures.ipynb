# All about apertures

This section provides a brief overview of the types of aperturees available in [`photutils`](https://photutils.readthedocs.io/en/stable/) and the aperture setetings that affect the measure of light. The following topics will be discussed:

+ Aperture shapes (e.g. circle, square, ellipse, custom)
+ Handling of partial pixels
+ Locating the centroid where the center of the aperture will be placed
