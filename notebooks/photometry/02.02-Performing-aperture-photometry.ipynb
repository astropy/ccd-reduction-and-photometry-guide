# Performing aperture photometry with [`photutils`](https://photutils.readthedocs.io/en/stable/) 

There a couple of different ways to perform aperture photometry in [`photutils`](https://photutils.readthedocs.io/en/stable/): using the [`aperture_photometry](https://photutils.readthedocs.io/en/stable/api/photutils.aperture.aperture_photometry.html#photutils.aperture.aperture_photometry) or the [`ApertureStats`](https://photutils.readthedocs.io/en/stable/api/photutils.aperture.ApertureStats.html#photutils.aperture.ApertureStats) class. While both produce identical photometry, they differ in what other information is available after the photometry and in how background subtraction is handled.

In this section we will use the same three images we did in source detection, with the method used to perform the photometry indicated in the title of the section.
