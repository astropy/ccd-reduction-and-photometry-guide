# Обработка overscan, trimming и вычитание bias

Bias в CCD камере представляет собой постоянное смещение, применяемое ко всем пикселям, чтобы при преобразовании напряжения в каждом пикселе в число это число всегда было положительным. В идеальной CCD bias был бы одинаковым для каждого пикселя и не изменялся бы со временем. На практике bias немного отличается для каждого пикселя и может изменяться на один-два отсчета от ночи к ночи или в течение ночи.

Bias *изображение* — это снимок, сделанный с закрытым затвором и нулевым временем экспозиции; представьте это как команду камере выполнить все обычные действия для подготовки электроники камеры к съемке изображения, а затем немедленно считать данные с CCD, как если бы вы сделали снимок.

*Overscan* — это часть CCD чипа, которая не подвергается воздействию света. В некоторых камерах это буквально несколько дополнительных строк и/или столбцов, которые не освещаются, в то время как в других overscan создается электронным способом. Он может быть полезен, если уровень bias изменяется в течение ночи.
