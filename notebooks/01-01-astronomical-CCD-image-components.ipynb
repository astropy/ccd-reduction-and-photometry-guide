# Понимание астрономического CCD изображения

Астрономическое изображение, подобное показанному ниже, по сути представляет собой двумерный
массив значений. В идеальном мире значение каждого пикселя (пиксель - это один
элемент массива) было бы прямо пропорционально количеству света, которое
попало на пиксель в течение времени, когда затвор камеры был открыт.

Но идеальный сценарий на самом деле не выполняется. Твердое понимание того,
*почему* значения пикселей не прямо пропорциональны свету, полезно перед
погружением в детали обработки изображений.

## Counts, фотоны и электроны

Число, хранящееся в необработанном астрономическом изображении прямо с телескопа, называется
Analog Digital Unit (ADU) или count, потому что внутри камера преобразует
аналоговое напряжение в каждом пикселе в численный count. Counts, представляющие интерес для
астронома, - это те, которые генерируются посредством фотоэлектрического эффекта, когда фотон
попадает на детектор. Число фотонов (или эквивалентно, электронов), которые достигают
пикселя, связано с counts в пикселе через gain (усиление).

Gain обычно предоставляется производителем камеры и может быть
измерен из комбинации bias и flat изображений (Howell 2002; p. 71).

**Обратите внимание**, что попытка преобразовать необработанный count изображения в фотоны/электроны путем
умножения на gain не будет иметь смысла, потому что необработанные counts включают
вклады из источников, отличных от света.

## Не все counts являются (интересным) светом

Существует несколько вкладов в counts в пикселе. Обработка изображений -
это, по сути, процесс удаления всех из них, за исключением тех, которые вызваны светом от
астрономического объекта:

+ Смещающее напряжение, называемое **bias**, применяется к CCD чипу для обеспечения того, чтобы
не было отрицательных counts во время считывания. Существуют небольшие вариации в значении
bias по всему чипу, и могут быть небольшие вариации в уровне bias
с течением времени.
+ Counts могут генерироваться в пикселе из-за теплового движения электронов в CCD;
охлаждение CCD уменьшает, но может не полностью устранить этот **dark current** (темновой ток). В
современных CCD dark current часто пренебрежимо мал, за исключением небольшой доли
пикселей. Dark current обычно указывается в электронах/секунду/пиксель, и
сильно зависит от температуры.
+ Существует **read noise** (шум считывания), присущий электронике CCD. Невозможно
устранить этот шум (он присутствует в каждом изображении, сделанном
камерой), но есть подходы к его минимизации. Read noise обычно
указывается в электронах, так как он может зависеть от температуры.
+ Часть света, получаемого телескопом, является рассеянным светом, исходящим от ночного
неба. Количество **sky background** (фона неба) зависит от полосы пропускания фильтра,
атмосферных условий и местных источников света.
+ Хотя CCD чип довольно мал, нередко **космические лучи** попадают
в чип, высвобождая заряд, который затем преобразуется в counts.

Все, что остается после удаления всех этих вещей, - это, в принципе, свет
от астрономических источников.

На практике существуют дополнительные сложности.

## CCD не идеальны

Существует ряд проблем, которые влияют на чувствительность CCD к свету,
некоторые из которых могут быть скорректированы, а некоторые - нет.

+ Виньетирование, затемнение изображений в углах, является обычным явлением и
корректируемым.
+ Пыль в оптическом пути, которая вызывает "пончики" или "червей" на изображении,
также является обычным явлением и корректируемой.
+ Вариации в чувствительности отдельных пикселей также обычны и
корректируемы.
+ Мертвые пиксели, которые не реагируют на свет, не могут быть скорректированы.

Коррекции **flat** пытаются удалить многие из этих эффектов. Идея состоит в том, чтобы
получить изображение чего-то, что равномерно освещено, как способ измерить вариации в
чувствительности (независимо от причины) и компенсировать их.
