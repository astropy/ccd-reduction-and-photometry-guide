# Подробнее о коррекции flat

## Обзор получения изображений для коррекции flat

Существует несколько способов получения изображений "flat":

+ Twilight flats — это изображения неба вблизи зенита, полученные на рассвете или закате.
+ Dome flats — это изображения внутренней части купола (обычно гладкой поверхности, а не самого купола), освещённой каким-либо источником света в куполе. Для небольших телескопов в качестве источника света можно использовать электролюминесцентную или светодиодную панель.
+ Sky flats составляются из нескольких science изображений.

В идеале изображения flat имеют достаточно высокие значения отсчётов (примерно половину от максимальных отсчётов детектора), так что *единственным* важным источником ошибки является пуассоновская ошибка, обусловленная светом в изображениях flat, и чтобы отношение сигнал/шум в этих изображениях было практически нулевым.

## Калибровка и комбинирование изображений flat

Процесс калибровки и комбинирования кадров flat в значительной степени одинаков независимо от используемого источника света.

Полезно думать о кадрах flat так же, как о science изображениях звёзд или галактик. Телескоп получает изображение источника света, поэтому bias и dark необходимо удалить из отдельных изображений.

При комбинировании изображений появляется новый шаг, который мы ещё не обсуждали: нормализация (также называемая масштабированием) откалиброванных кадров flat к общему среднему или медианному значению перед их комбинированием. Как в sky, так и в twilight flats освещение естественным образом меняется от кадра к кадру. Если изображения не масштабированы к общему значению перед комбинированием, то те, которые были получены при более ярком небе, будут неуместно доминировать в результате. Dome flats должны быть, в принципе, совершенно стабильными без временной вариации их освещённости. На практике каждый источник света варьируется на каком-то уровне; если вы пытаетесь скорректировать 1% различия в освещённости, то 1% колебания источника света имеют значение.

Обычно среднее или медианное значение масштабируется до 1.0 перед комбинированием, так что когда science изображения делятся на откалиброванные, скомбинированные flats, значения science изображения не слишком сильно изменяются.
